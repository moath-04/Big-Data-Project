```python
df_clean = df.drop_duplicates().copy()
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Removed duplicates:", len(df) - len(df_clean))
```

```python
# print(df_clean["Weather"].unique())
print(df_clean["Weather"].value_counts(dropna=False))
```

```python
df_clean["Weather"] = df_clean["Weather"].str.strip().str.lower()

print(df_clean["Weather"].value_counts(dropna=False))
```

```python
missing_values =df_clean.isna().sum()
missing_values = missing_values[missing_values > 0]
print(missing_values)
```

```python
print(df_clean["Recorded_Speed_kmh"].describe())

print("\nTop 20 highest speeds:")
print(df_clean["Recorded_Speed_kmh"].nlargest(20))

print("\nSpeeds above 260 km/h:")
print((df_clean["Recorded_Speed_kmh"] > 260).sum())
```

```python
# تغيير السرعة الغير منطقية الى متوسط السرعات
valid_speeds = df_clean.loc[
    df_clean["Recorded_Speed_kmh"] != 999,
    "Recorded_Speed_kmh"
]

mean_speed = valid_speeds.mean()

print("Mean of valid speeds:", mean_speed)

df_clean.loc[
    df_clean["Recorded_Speed_kmh"] == 999,
    "Recorded_Speed_kmh"
] = mean_speed

print("Remaining 999 values:", (df_clean["Recorded_Speed_kmh"] == 999).sum())
print("Total rows:", len(df_clean))
```

```python
print(df_clean["Accident_Occurred"].value_counts())
accidents_df = df_clean[df_clean["Accident_Occurred"] == 1].copy()
```

```python
# تغيير نوع التاريخ من اوبجكت لdatetime
accidents_df["Timestamp"] = pd.to_datetime(accidents_df["Timestamp"],errors="coerce")

print("New data type:", accidents_df["Timestamp"].dtype)

print("Invalid timestamps:", accidents_df["Timestamp"].isna().sum())
```

```python
accidents_df = accidents_df.sort_values( by="Timestamp",ascending=True).reset_index(drop=True)

print("Total accidents:",len(accidents_df))
print("First accident:",accidents_df["Timestamp"].iloc[0])
print("Last accident:", accidents_df["Timestamp"].iloc[-1])
print("Is sorted:", accidents_df["Timestamp"].is_monotonic_increasing)
```

```python
processed_dir = project_dir / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

# Save cleaned accident records
output_file = processed_dir / "accidents_clean.csv"
accidents_df.to_csv(output_file, index=False)

print("File saved:", output_file.exists())
print("File location:", output_file)
print("Total saved accidents:", len(accidents_df))
```

## Data Cleaning Summary

### Dataset Overview
- Original dataset: 200,000 records and 29 columns.
- Dataset type: Synthetic traffic and emergency response data.
- Final dataset: 8,213 accident records.

### Data Cleaning Steps

**1. Duplicate Removal**
- Identified and removed 597 fully duplicated records.
- Remaining records: 199,403.

**2. Weather Standardization**
- Standardized weather categories using `strip()` and `lower()`.
- Combined inconsistent variations of Rain into one category.
- Final categories: clear, rain, snow, and fog.

**3. Missing Values**
- Visibility_m: 5,079 missing values.
- Driver_Experience_yrs: 3,986 missing values.
- Missing values were retained for later consideration during model preprocessing.

**4. Invalid Speed Handling**
- Identified 104 records with a recorded speed of 999 km/h.
- Replaced these values with the mean calculated after excluding 999 km/h readings.
- Replacement mean: approximately 49.1066 km/h.
- No records were removed during this operation.
- Mean imputation provides estimated values rather than actual recorded speeds and may affect speed-related statistics.
- Recorded speed is not among the seven selected features for the prediction model.

**5. Accident Filtering**
- Selected records where `Accident_Occurred == 1`.
- Retained 8,213 accident records.

**6. Chronological Sorting**
- Converted Timestamp to datetime.
- Verified that no invalid timestamps remained.
- Sorted accidents from oldest to newest.

### Final Validation
- Final dataset shape: 8,213 rows × 29 columns.
- All records represent accidents.
- No fully duplicated records remain.
- Timestamps are sorted chronologically.
- No speed values equal to 999 km/h.

### Output
`data/processed/accidents_clean.csv`
